```python
# Exercise 5: Sorting Data
data = {
    "trial_id": ["NCT001", "NCT002", "NCT003"],
    "sponsor_name": ["Pfizer", "AstraZeneca", "GSK"],
    "start_year": [2020, 2021, 2019]
}

df = pd.DataFrame(data)

df.sort_values(
    ["start_year", "sponsor_name"],
    ascending=[False, True]
)
```

```python
#Exercise 6: Dates
# real-clinical trials data will use full dates such as 2021-05-14

date_data={
    "trial_id": ["NCT001", "NCT002", "NCT003"],
    "start_date": ["2020-03-15", "2021-07-01", "2019-11-20"]
}
date_df=pd.DataFrame(date_data)
print(date_df)

date_df.info()

date_df["start_date"] = pd.to_datetime(date_df["start_date"])

date_df.info()
date_df["start_year"] = date_df["start_date"].dt.year
date_df["start_month"] = date_df["start_date"].dt.month

date_df

```

```python
#7 Exercise 7: Data Types

type_data = {
    "trial_id": ["NCT001", "NCT002", "NCT003"],
    "enrollment": ["120", "250", "180"],
    "industry_sponsored": [True, False, True]
}

type_df = pd.DataFrame(type_data)

type_df.info()
#enrollment is inside the quotes so it is being treated as a string (object) data type. To perform numerical operations on this column, we need to convert it to an integer data type.

type_df["enrollment"] = type_df["enrollment"].astype(int)
type_df.info()

print(type_df["enrollment"].dtype)

#key idea is that .astype(int) means convert this column into integer
```

```python
#Exercise 8: Groupby

group_data = {
    "sponsor_name": ["Pfizer", "Pfizer", "GSK", "GSK", "AstraZeneca"],
    "enrollment": [120, 180, 200, 150, 250]
}

group_df = pd.DataFrame(group_data)

group_df

group_df.groupby("sponsor_name")["enrollment"].sum()

group_df.groupby("sponsor_name")["enrollment"].mean()
group_df.groupby("sponsor_name")["enrollment"].count()
group_df.groupby("sponsor_name", as_index=False)["enrollment"].sum()

#as_index keeps sponsor_name as a normal column which is easier to work with latr one

# give the average enrolment
group_df.groupby("sponsor_name", as_index=False)["enrollment"].mean()
```

```python
#Exercise 9: Merge

trials_df = pd.DataFrame({
    "trial_id": ["NCT001", "NCT002", "NCT003"],
    "sponsor_name": ["Pfizer", "GSK", "AstraZeneca"]
})

countries_df = pd.DataFrame({
    "trial_id": ["NCT001", "NCT002", "NCT004"],
    "country": ["United Kingdom", "United States", "France"]
})

merged_df = trials_df.merge(
    countries_df,
    on="trial_id",
    how="inner"
)

merged_df

# only returns NCT001 and NCT002 because they are the only trial_ids that exist in both DataFrames. The trial_id NCT004 from countries_df is not included in the result because it does not have a matching trial_id in trials_df.

left_merged_df = trials_df.merge(
    countries_df,
    on="trial_id",
    how="left"
)

left_merged_df
```
